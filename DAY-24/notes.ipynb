https://leetcode.com/problems/jewels-and-stones/

```python
jewels = "aA" 
stones = "aAAbbbb"
ans=0
for i in stones:
    if i in jewels:
        ans+=1
print(ans)
```

```python
jewels = "aA" 
stones = "aAAbbbb"

# for stones am going with frequency count
rec={}
for i in stones:
    if i in rec:
        rec[i]+=1
    else:
        rec[i]=1
print(rec)
```

```python
jewels = "aA" 
stones = "aAAbbbb"

# for stones am going with frequency count
rec={}
for i in stones:
    if i in rec:
        rec[i]+=1
    else:
        rec[i]=1
print(rec)
ans=0
for i in jewels:
    if i in rec:
        ans+=rec[i]
print(ans)
```

```python
key = "the quick brown fox jumps over the lazy dog"
message = "vkbs bs t suepuv"
rec={}
temp=97
for i in key:
    if i!=' ' and i not in rec:
        rec[i]=chr(temp)
        temp+=1
print(rec)
```

```python
key = "the quick brown fox jumps over the lazy dog"
message = "vkbs bs t suepuv"
rec={}
temp=97
for i in key:
    if i!=' ' and i not in rec:
        rec[i]=chr(temp)
        temp+=1
print(rec)
output=""
for i in message:
    if i!=' ':
        output+=rec[i]
    else:
        output+=' '
print(output)
```
