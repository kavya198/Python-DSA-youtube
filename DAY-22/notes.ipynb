```python
nums = [2,2,1,1,1,2,2]
n = len(nums)
rec={}
for key in nums:
    if key in rec:
        rec[key]+=1
    else:
        rec[key]=1
ans = -1
temp = n/2
for key in rec:
    if rec[key]> temp:
        ans = key
        break
print(ans)
```
