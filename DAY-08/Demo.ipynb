```python
#def finalValueAfterOperations(self, operations: List[str]) -> int:
operations = ["X++","++X","X++"]
X=0
for i in operations:
        if i == "X++" or i == "++X":
            X+=1
        if i == "X--" or i == "X--":
            X-=1
print(X)
```

```python
jewels = 'aA'
stones = 'aAAbbbb'
ans = 0
for i in range(len(jewels)):
    chi = jewels[i]
    for j in range(len(stones)):
        chj = stones[j]
        if chi == chj:
            ans+=1
print(ans)
```

```python
jewels = 'aA'
stones = 'aAAbbbb'
ans=0
dup=[]
for i in jewels:
    for j in stones:
        if i != dup:
            ans+=1
            dup.append(i)

            break
print(ans)
```
