```python
s="aabaa"
n=len(s)//2
m=n*2
valid = True
for i in range(n):
    left=s[i]
    right=s[m-i-1]
    if left != right:
        break
    print(left,right)

if left ==right:
    print("Yes Its palindrome")
else:
    print("No Its not palindrome")
```
