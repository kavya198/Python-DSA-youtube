```python
n = 4

# --- Code for the First Pattern (The Right-Angle Triangle) ---
#print("First Pattern:")
for i in range(n):
    # The number of stars needed is i + 1
    num_stars = i + 1
    
    for j in range(num_stars):
        print(j+1, end=' ')
    
    print() # Move to the next line

# Add some empty lines for separation
#print()
#print("Second Pattern:")
#print()

# --- Code for the Second Pattern (The Inverted Right-Angle Triangle) ---
for i in range(n):
    # The number of stars needed is n - i (4 in first row, 3 in second, etc.)
    num_stars = n - i
    
    for j in range(num_stars):
        print(i+1, end=' ')
        
    print() # Move to the next line
```
