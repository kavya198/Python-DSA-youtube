```python
arr = [5,4,3,2]
n=len(arr)
for i in range(n):
    for j in range(i,n):
        for k in range(i,j+1):
            print(arr[k],end=" ")
        print()
```

```python
from collections import Counter

items = ["apple", "banana", "apple", "orange", "banana", "banana"]

counts = Counter(items)
# Filter items that appear more than once
duplicates = [item for item, count in counts.items() if count > 1]

print(duplicates)  # Output: ['apple', 'banana']
print(counts)       # Output: Counter({'banana': 3, 'apple': 2, 'orange': 1})
```

```python
items = [1, 2, 3, 2, 4, 5, 1, 6]

seen = set()
duplicates = set()

for item in items:
    if item in seen:
        duplicates.add(item)
    else:
        seen.add(item)

print(duplicates)  # Output: {1, 2}
```

```python
# Line 1: Import defaultdict to automatically initialize keys that do not exist yet.
from collections import defaultdict
# Line 2: Import List from typing to provide clear type hints for code readability.
from typing import List

# Line 4: Define the function which accepts a list of string logs and returns a dictionary.
def aggregate_api_traffic(logs: List[str]) -> dict:
    # Line 6: Create a tracking map. Missing keys will automatically default to an integer (0).
    endpoint_counts = defaultdict(int)
    
    # Line 8: Start a loop to process every individual log string inside the list.
    for log_entry in logs:
        # Line 10: Strip trailing whitespaces and split the line by spaces into an array of words.
        parts = log_entry.strip().split()
        
        # Line 12: Ensure the line has at least 3 components to prevent IndexError crashes.
        if len(parts) >= 3:
            # Line 14: Extract the HTTP method (e.g., 'POST') which is the second element (index 1).
            method = parts[1]
            # Line 16: Extract the API URL route (e.g., '/v1/trades') which is the third element (index 2).
            path = parts[2]
            
            # Line 19: Combine method and path using an f-string to form a unique lookup key string.
            endpoint_key = f"{method} {path}"
            # Line 21: Increment the hit counter for this specific endpoint by 1.
            endpoint_counts[endpoint_key] += 1
            
    # Line 24: Convert the standard defaultdict back into a clean, regular Python dictionary.
    return dict(endpoint_counts)

# Line 27: Standard Python boilerplate ensuring the script only runs if executed directly.
if __name__ == "__main__":
    # Line 29: Create an array of mock system logs matching real-world environment formats.
    sample_logs = [
        "10:00:01 GET /v1/health 200",
        "10:00:02 POST /v1/trades 201",
        "10:00:03 GET /v1/health 200",
        "10:00:04 GET /v1/balances 200"
    ]
    # Line 36: Execute the function with the mock data and display the result on screen.
    print(aggregate_api_traffic(sample_logs))
```

```python
s = "Hello, World!"
def reverse_string(s):
    chars = list(s)
    left, right = 0, len(chars) - 1
    while left < right:
        chars[left], chars[right] = chars[right], chars[left]
        left += 1
        right -= 1
    return "".join(chars)
print(reverse_string(s))  # Output: !dlroW ,olleH
```

```python
s="Hi"
def reverse_string_1(s):
    result = ""
    for char in s:
        result = char + result
    return result
print(reverse_string_1(s))  # Output: iH    
```
