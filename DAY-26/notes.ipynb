# Sub Array

```python
arr = [5,4,3,2]
n=len(arr)
for i in range(n):
    for j in range(i,n):
        for k in range(i,j+1):
            print(arr[k],end=" ")
        print()
```

```python
arr = [5,4,3,2]
n=len(arr)
for i in range(n):
    for j in range(i,n):
        temp=[]
        for k in range(i,j+1):
            temp.append(arr[k])
        print(temp)
```

```python
arr = [5,4,3,2]
n=len(arr)
grp_arr = []
for i in range(n):
    for j in range(i,n):
        temp=[]
        for k in range(i,j+1):
            temp.append(arr[k])
        grp_arr.append(temp)
print(grp_arr)
```
