```python
nums = [1,2,3,1,1,3]
n = len(nums)
for i in range(n):
    for j in range(i+1,n):
        print(i,j)
```

```python
nums = [1,2,3,1,1,3]
n = len(nums)
for i in range(n):
    for j in range(i+1,n):
        print(nums[i],nums[j])
```

```python
nums = [1,2,3,1,1,3]
n = len(nums)
ans = 0
for i in range(n):
    for j in range(i+1,n):
        if nums[i] == nums[j]:
            ans+=1
print(ans)
```

```python
nums = [1,2,3,1,1,3]

rec = {}
for i in nums:
    if i in rec:
        rec[i]+=1
    else:
        rec[i]=1
print(rec)
```

```python
nums = [1,2,3,1,1,3]

## frequency count
rec = {}
for i in nums:
    if i in rec:
        rec[i]+=1
    else:
        rec[i]=1
print(rec)


ans=0
for i in rec:
    n=rec[i]
    temp=n*(n-1)//2
    ans=ans+temp
print(ans)
```
